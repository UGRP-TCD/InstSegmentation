```python
import cv2
import numpy as np

def yolo_resize(image, target_size=640):
    # 원본 이미지 크기 가져오기
    h, w, c = image.shape
    
    # 가장 긴 변이 target_size가 되도록 비율 계산
    scale = target_size / max(h, w)
    
    # 새로운 이미지 크기 계산
    new_w, new_h = int(w * scale), int(h * scale)
    
    # 이미지 리사이즈 (비율 유지)
    resized_image = cv2.resize(image, (new_w, new_h), interpolation=cv2.INTER_LINEAR)
    
    # 새로운 이미지를 위한 빈 캔버스 생성 (패딩 포함)
    canvas = np.full((target_size, target_size, 3), 128, dtype=np.uint8)  # 패딩은 중간 회색 (128)
    
    # 이미지가 중앙에 위치하도록 좌상단 좌표 계산
    top_left_x = (target_size - new_w) // 2
    top_left_y = (target_size - new_h) // 2
    
    # 캔버스에 리사이즈된 이미지 복사
    canvas[top_left_y:top_left_y+new_h, top_left_x:top_left_x+new_w] = resized_image
    print(new_h,new_w)
    
    return canvas

# 예제 사용
image = cv2.imread('test_img.jpg')  # 이미지를 불러옵니다.
resized_image = yolo_resize(image, target_size=640)

# 결과를 확인하려면 다음 코드를 실행하세요.
cv2.imshow('YOLO Resized Image', resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
